```python
# barcha kerakli bo'lgan kutubxonalarni yuklab olamiz
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
import pandas as pd
from sklearn import preprocessing
```

```python
from sklearn.preprocessing import LabelEncoder, StandardScaler, OneHotEncoder, OrdinalEncoder
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
```

```python
titanik_df = pd.read_csv('/content/titanic.csv',) #header = None, na_values='?')
```

```python
titanik_df.sample(5) # tahminiy 5ta qatorni oldik hatolar va jadvalni uzini kurish uchun
```

```python
titanik_df.info() # tipini korib olishimiz kerak
```

```python
titanik_df['Embarked'].unique() #har bir bulimdi ichida nima hatolari borligini tekshiramiz masalan bitta soz har xilcha qilib yozilgan bulishi mumkin katta har kichik harf degande
```

```python
le = LabelEncoder()  #faqat ir ulchovli massivlar bilan ishlidi yomon tomoni shunda - yaxshisi esa buni bir biriga kopaytirib bolmaydi
oe = OrdinalEncoder() # bu bizga qulay sababi pipiline ichida bir nechta bulimlar bilan ishlashda - yomoni bir biriga kupaytirsa hato bolib ketadi
```

```python
oe.fit_transform(titanik_df[['Sex', 'Ticket']]) # databazadagi 'Sex', 'Ticket' bolimlarini raqamli ustunga aylantirib matrix kurinishida qaytaradi
```

```python
X = titanik_df.drop(columns=['Survived', 'PassengerId']) # shu ustun bizga kerak va buni x dan chiqarib tashimizda, y ga qushamiz
y = titanik_df['Survived'] # faqat shu ustunni olamiz
```

```python
X.dtypes
```

```python
cat_cols_names = X.select_dtypes('object').columns.to_list() # X dagi berilgan ustunlarni faqat matnlilarini (stirng, object) olamiz
num_cols_names = X.select_dtypes('number').columns.to_list() # X da berilgan ustunlarni faqat numeric (float64, int64)larini olamiz
num_cols_names
```

```python
cat_process = Pipeline([
    ('impute', SimpleImputer(strategy='most_frequent')),
    ('ore', OrdinalEncoder())
])
num_process = Pipeline([
    ('impute', SimpleImputer(strategy='mean')),
    ('scaler', StandardScaler())
])
preprocess = ColumnTransformer([
    ('cat', cat_process, cat_cols_names),
    ('num', num_process, num_cols_names)
]).set_output(transform='pandas') # bunda barcha cols larni birlashtirib hatoliklarini tugirlab chiqyapmiz va bush joylarni ham toldirib ketyapmiz
```

```python
X_data = preprocess.fit_transform(X) # bu jadvalda uje hamma nomli qatorlar raqamliga utkan boladi
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
```

```python
X_train, X_test, y_train, y_test = train_test_split(X_data, y, test_size=0.2)
print(X_train.shape, X_test.shape, y_train.shape, y_test.shape, sep='\n') # modelni trenin qildirish uchun ularni tayyorlab oldik
```

```python
model = LogisticRegression(max_iter=750)
model # hali trenin qildirilmagan model
```

```python
model.fit(X_train, y_train)
```

```python
y_pred = model.predict(X_test)
y_probs = model.predict_proba(X_test)[:, 1]
```

```python
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay, accuracy_score, recall_score, precision_score, f1_score, classification_report
import matplotlib.pyplot as plt
```

```python
cm = confusion_matrix(y_test, y_pred)
```

```python
ConfusionMatrixDisplay.from_predictions(y_test, y_pred)
plt.show()
```

```python
display_cm = ConfusionMatrixDisplay(confusion_matrix=cm)
display_cm.plot()
plt.show()
```

```python
threshold = 0.7
y_pred_threshold_07 = np.array([1 if x >= threshold else 0 for x in y_probs])
```

```python
threshold = 0.35
y_pred_threshold_035 = np.array([1 if x >= threshold else 0 for x in y_probs])
```

```python
precision = (cm[1, 1]) / (cm[1, 1] + cm[0, 1])
print("Pricision_score: ", precision)
print(precision_score(y_test, y_pred))
```

```python
recall = (cm[1, 1]) / (cm[1, 1] + cm[1, 0])
print("Recall_score: ", recall)
print(recall_score(y_test, y_pred))
```

```python
print("Accuracy_score: ", (cm[0, 0] + cm[1, 1]) / (cm[1, 1] + cm[0, 1] + cm[0, 0] + cm[1, 0]))
print(accuracy_score(y_test, y_pred))
```

```python
print("F1_score: ", 2*precision*recall / (precision + recall))
print(f1_score(y_test, y_pred))
```

```python
print('Accuracy_score', (cm[0, 0]) + cm[1, 1] / (cm[1, 1] + cm[0, 1] + cm[0, 0] + cm[1,0]))
```

```python
print(classification_report(y_test, y_pred))
```

```python
print(classification_report(y_test, y_pred_threshold_07))
```

```python
print(classification_report(y_test, y_pred_threshold_035))
```
